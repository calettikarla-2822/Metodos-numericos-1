```python
import numpy as np
import pandas as pd

def interpolacion_neville(x_eval, x_vals, y_vals):
    """
    Calcula la tabla de interpolación de Neville.
    x_eval: Punto donde se evalúa el polinomio (x)
    x_vals: Lista de nodos x0, x1, ..., xn
    y_vals: Lista de f(x0), f(x1), ..., f(xn)
    """
    n = len(x_vals)
    # Iniciar la matriz Q con ceros
    Q = np.zeros((n, n))

    # La primera columna son los valores de f(x)
    for i in range(n):
        Q[i, 0] = y_vals[i]

    # Calcular el resto de la tabla usando la fórmula de Neville
    for i in range(1, n):
        for j in range(1, i + 1):
            # Fórmula de Neville
            term1 = (x_eval - x_vals[i - j]) * Q[i, j - 1]
            term2 = (x_eval - x_vals[i]) * Q[i - 1, j - 1]
            Q[i, j] = (term1 - term2) / (x_vals[i] - x_vals[i - j])

    return Q

print("Función de Neville cargada correctamente.")
```

```python
# --- EJEMPLO 2 ---
# Nodos x
x_nodos = [1.0, 1.3, 1.6, 1.9, 2.2]
# Valores f(x) (primera columna de la tabla)
y_valores = [0.7651977, 0.6200860, 0.4554022, 0.2818186, 0.1103623]
# Valor de x a evaluar
x_target = 1.5

print(f"Calculando interpolación para x = {x_target}...\n")

# Calcular la tabla
tabla_Q = interpolacion_neville(x_target, x_nodos, y_valores)

print("Cálculo terminado. Ve a la Celda 3 para ver el paso a paso.")
```

```python
print(f"--- CÁLCULOS ---\n")

print("Aproximaciones de primer grado:")
print(f"Q1,1(1.5) = {tabla_Q[1,1]:.7f}")
print(f"Q2,1(1.5) = {tabla_Q[2,1]:.7f}")
print(f"Q3,1(1.5) = {tabla_Q[3,1]:.7f}")
print(f"Q4,1(1.5) = {tabla_Q[4,1]:.7f}\n")

print("Aproximaciones de segundo grado:")
print(f"Q2,2(1.5) = {tabla_Q[2,2]:.7f}")
print(f"Q3,2(1.5) = {tabla_Q[3,2]:.7f}")
print(f"Q4,2(1.5) = {tabla_Q[4,2]:.7f}\n")

print(f"El valor final de la interpolación es P(1.5) = Q4,4 = {tabla_Q[4,4]:.7f}")
```

```python
# --- Tabla 3.6 ---
datos_tabla = []
for i in range(len(x_nodos)):
    fila = [x_nodos[i]]
    # Añadir los valores calculados hasta la diagonal (j <= i)
    for j in range(i + 1):
        fila.append(tabla_Q[i, j])

    while len(fila) < len(x_nodos) + 1:
        fila.append("")
    datos_tabla.append(fila)

columnas = ["x_i", "Q_i,0", "Q_i,1", "Q_i,2", "Q_i,3", "Q_i,4"]
df_resultado = pd.DataFrame(datos_tabla, columns=columnas)

# Redondear a 7 decimales y dejar vacíos los ceros que sobran de forma segura
for col in columnas[1:]:
    df_resultado[col] = df_resultado[col].apply(lambda x: f"{x:.7f}" if isinstance(x, (int, float, np.number)) and x != 0 else (x if isinstance(x, str) else ""))

print("--- Tabla 3.6 (Resultados) ---")
print(df_resultado.to_string(index=False))

# --- Añadiendo el nodo x5 = 2.5 ---
print("\n--- Añadiendo el nodo x5 = 2.5 ---")
x_nodos_ext = x_nodos + [2.5]
y_valores_ext = y_valores + [-0.0483838]
tabla_ext = interpolacion_neville(x_target, x_nodos_ext, y_valores_ext)

print(f"La nueva aproximación Q5,5(1.5) = {tabla_ext[5,5]:.7f} (Coincide con el libro).")
```
